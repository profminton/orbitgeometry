## Orbital elements and generic conic section geometric parameters
$\mu$
: Gravitational parameter: $\mu=GM$

$M$
: Mass of central body

$\mathbf{r}$
: Orbit radius vector

$\mathbf{v}$
: Orbit velocity vector

$\mathbf{h}$
: Specific angular momentum vector: $\mathbf{h}=\mathbf{r}\times\mathbf{v}$

$\mathcal{E}$ 
: Specific energy (the *vis-viva integral*): $\mathcal{E}=\frac{v^2}{2}-\frac{\mu}{r}$

$\phi$
: Flight path angle: $\phi = \cos^{-1}\left(\frac{h}{rv}\right)$

The equation of a conic section in polar coordinates aligned with the focus-periapsis axis
: $r = \frac{p}{1 + e \cos(f)}$

$p$
: Semilatus rectum: $p=h^2/\mu$

$e$:
Eccentricity

$f$
: True anomaly

$\varpi$
: Longitude of periapsis (orientation of the orbit with respect to an independent reference frame)

$F$ 
: Primary focus

$F'$
: Empty focus

## Parameters for the elliptical orbit
$T$
: Period of the orbit (Kepler's 3rd law): $T^2=\frac{4\pi^2}{\mu}a^3$ 

$n$
: Mean motion: $n^2=\frac{\mu}{a^3}$

$p$
: Semilatus rectum: $p = a\left(1-e^2\right)$

$a$
: Semimajor axis (half of the long axis length). 

$b$ 
: Semiminor axis (half of the short axis length): $b=a\sqrt{1-e^2}$

$e$
: Eccentricity (<1): $e=\sqrt{1-b^2/a^2}$

$c$
: Linear eccentricity: $c=ae$

$r_p$
: Periapsis (closest point in the orbit): $r_p = a\left(1-e\right)$

$r_a$
: Apoapsis (furthest point in the orbit): $r_a = a\left(1+e\right)$

$\mathcal{E}$
: Specific energy (<0): $\mathcal{E} = -\frac{\mu}{2a}$

$h$
: Specific angular momentum magnitude: $h=na^2\sqrt{1-e^2}=\sqrt{\mu a\left(1-e^2\right)}$

## Parameters for the hyperbolic trajectory
*Note: The choice of sign of $a$ for hyperbolic trajectories is arbitrary. Most commonly $a<0$ for the hyperbolic trajectory, however, Murray & Dermott use $a>0$, which is what I have adopted below.*

$p$
: Semilatus rectum: $p = a\left(e^2-1\right)$

$a$
: Semimajor axis (the distance between the periapsis and mid-focus point). 

$e$
: Eccentricity (>1): $e=\sqrt{1+b^2/a^2}$

$b$ 
: Semiminor axis (impact parameter): $b=a\sqrt{e^2-1}$

$c$
: Linear eccentricity: $c=ae$

$r_p$
: Periapsis (closest point in the orbit): $r_p = a\left(e-1\right)$

$\psi$
: Deflection angle: $\sin\frac{\psi}{2}= e^{-1}$

$\mathcal{E}$
: Specific energy (>0): $\mathcal{E} = \frac{\mu}{2a}$

$v_{\infty}$
: Hyperbolic excess velocity: $v_\infty^2=\frac{\mu}{a}$, where $v_\infty = v\left|_{r\rightarrow\infty}\right.$

$h$
: Specific angular momentum magnitude: $h=bv_\infty=\sqrt{\mu a\left(e^2-1\right)}$

$y_a(x)$
: Asymptote lines: $y_a(x) = \pm\frac{b}{a}x_a$, where $x$ and $y_a$ are in a coordinate system centered between the foci.
